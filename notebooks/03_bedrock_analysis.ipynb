## Bedrock Analysis Results

Results from running `src/bedrock_inference.py` using Claude Sonnet 4.6 via AWS Bedrock. Ran on a sample of 20 reviews to verify the pipeline functions correctly before scaling up.


```python
import boto3
import pandas as pd
from io import BytesIO

s3 = boto3.client('s3')
obj = s3.get_object(Bucket="candrews-sentiment-pipeline", Key="results/bedrock_results.parquet")
df = pd.read_parquet(BytesIO(obj['Body'].read()))

label_map = {0: "negative", 1: "positive"}
df["correct"] = df["sentiment"] == df["original_label"].map(label_map)
print(df[["sentiment", "original_label", "correct"]].head(20))
print(f"\nAccuracy: {df['correct'].mean():.2%}")
```
